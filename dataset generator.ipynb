```python
import pandas as pd
import random
import numpy as np
from faker import Faker

# Initialiser Faker pour les noms français/arabes
fake = Faker()

# Élargir la liste des noms de famille togolais (sources réelles)
last_names = [
    "Agbeko", "Amouzou", "Gbati", "Akakpo", "Koffi", "Lare", "Ali", "Mensah", "Kombate",
    "Togbui", "Lamboni", "Assogba", "Douti", "Edoh", "Sossou", "Kolani", "Adjahoui", "Badjare",
    "Balogou", "Bawa", "Boko", "Dagbovie", "Dake", "Douti", "Eklou", "Foli", "Gaba", "Gadou",
    "Hola", "Kadanga", "Kokoroko", "Loumon", "Mawoula", "N'diaye", "Ocloo", "Padonou", "Quashie",
    "Sakpa", "Tete", "Vovor", "Walla", "Yawa", "Zinsou", "Abalo", "Bekale", "Dagba", "Foly", "Homawoo"
]

# Prénoms masculins étendus (togolais, français, arabes)
male_first_names = [
    # Togolais
    "Koffi", "Kodjo", "Komlan", "Kokou", "Kossi", "Boukari", "Kpatcha", "Sena", "Tchalla", 
    "Elie", "Atsu", "Kokouvi", "Koffitse", "Kossivi", "Têtê", "Akakpo", "Koku", "Komi", "Mawuli",
    "Koffigan", "Kossigan", "Lantame", "Dzigbodi", "Kondo", "Togbe", "Agbesi", "Kafui", "Kekeli",
    
    # Français/Internationaux
    "Jean", "Paul", "Marc", "Pierre", "Luc", "Thomas", "David", "Michel", "André", "Philippe",
    "Serge", "Claude", "Roger", "Yves", "Bernard", "Christian", "Daniel", "Éric", "Patrick",
    
    # Arabes
    "Mohamed", "Amadou", "Abdoulaye", "Ibrahim", "Ousmane", "Moussa", "Ali", "Mahamadou", "Idriss",
    "Boubacar", "Souleymane", "Ismaël", "Harouna", "Seydou", "Tidjani", "Aboubacar", "Karim"
]

# Prénoms féminins étendus (togolais, français, arabes)
female_first_names = [
    # Togolais
    "Ama", "Afi", "Adjoa", "Essi", "Yawa", "Abiba", "Ajara", "Sena", "Mawussi", "Kafui",
    "Akofa", "Dzigbodi", "Kekeli", "Mawunyega", "Sika", "Afiwa", "Akossiwa", "Essohana", "Komi",
    "Lola", "Nyonuku", "Têtê", "Yawo", "Zéwé", "Araba", "Dede", "Foli", "Kafui", "Mawufemor",
    
    # Français/Internationaux
    "Marie", "Jeanne", "Anne", "Sophie", "Catherine", "Françoise", "Élisabeth", "Monique", "Chantal",
    "Nathalie", "Isabelle", "Sylvie", "Martine", "Christine", "Brigitte", "Valérie", "Sandrine",
    "Julie", "Stéphanie", "Laurence", "Virginie", "Caroline", "Hélène", "Christelle",
    
    # Arabes
    "Fatoumata", "Aminata", "Mariam", "Aïssatou", "Hawa", "Kadiatou", "Ramatoulaye", "Oumou", "Fanta",
    "Salamata", "Adama", "Kadidia", "Djeneba", "Assetou", "Hadiza", "Zahra", "Aïcha", "Nafissatou"
]

# Terminaisons typiques pour améliorer la reconnaissance
male_suffixes = ["ou", "d", "i", "an", "en", "er", "in", "o", "ui", "vi"]
female_suffixes = ["a", "e", "é", "ie", "we", "ya", "ba", "na", "ta", "wa"]

# Générer des prénoms composés
def generate_compound_name(name_list, gender):
    """Génère des prénoms composés réalistes"""
    if random.random() < 0.25:  # 25% de prénoms composés
        separator = random.choice(["-", " "])
        return separator.join(random.sample(name_list, 2))
    return random.choice(name_list)

# Générer le dataset
total_rows = 10000
data = []

for _ in range(total_rows):
    # Choisir le genre (55% masculin comme dans la démographie)
    gender = 1 if random.random() < 0.55 else 0
    
    # Sélectionner le prénom adapté
    first_name = generate_compound_name(
        male_first_names if gender == 1 else female_first_names, 
        gender
    )
    
    # Ajouter des suffixes caractéristiques occasionnellement
    if random.random() < 0.4:
        if gender == 1:
            suffix = random.choice(male_suffixes)
            # Ne pas ajouter de suffixe si le prénom se termine déjà par une voyelle
            if first_name[-1] not in ['a', 'e', 'i', 'o', 'u', 'é']:
                first_name += suffix
        else:
            suffix = random.choice(female_suffixes)
            if first_name[-1] not in ['a', 'e', 'i', 'o', 'u', 'é']:
                first_name += suffix
    
    # Construire le nom complet
    full_name = f"{random.choice(last_names)} {first_name}"
    
    # Ajouter quelques noms avec deux prénoms
    if random.random() < 0.15:
        second_name = generate_compound_name(
            male_first_names if gender == 1 else female_first_names, 
            gender
        )
        full_name = f"{full_name} {second_name}"
    
    data.append({"full_name": full_name, "gender": gender})

# Créer le DataFrame
df = pd.DataFrame(data)

# Sauvegarder dans un fichier CSV
file_path = "./noms_prenoms_togo_10000.csv"
df.to_csv(file_path, index=False)

print(f"Dataset généré avec {total_rows} lignes : {file_path}")
```
